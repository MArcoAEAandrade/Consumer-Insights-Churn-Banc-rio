```python
%sql
SELECT 
    CASE 
        WHEN provincia_origem = 'Tỉnh khác' THEN 'Outras Províncias (Tỉnh khác)'
        ELSE 'Grandes Centros / Capitais'
    END AS grupo_regiao,
    churn,
    COUNT(*) AS total_clientes,
    ROUND(AVG(saldo), 2) AS media_saldo,
    ROUND(AVG(renda_mensal), 2) AS media_renda,
    ROUND(AVG(num_servicos), 2) AS media_servicos,
    ROUND(SUM(CASE WHEN membro_ativo = 'Inativo' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS pct_inativos
FROM workspace.default.bank_churn_dataset_pt
GROUP BY 
    CASE 
        WHEN provincia_origem = 'Tỉnh khác' THEN 'Outras Províncias (Tỉnh khác)'
        ELSE 'Grandes Centros / Capitais'
    END,
    churn;
```

```python
%sql
SELECT
    provincia_origem,
    COUNT(*) AS total_clientes,
    SUM(churn) AS total_churn,
    ROUND(SUM(churn) * 100.0 / COUNT(*), 2) AS taxa_churn_pct
FROM workspace.default.bank_churn_dataset_pt
WHERE segmento_cliente = 'Mass'
  AND nivel_fidelidade = 'Bronze'
GROUP BY provincia_origem
HAVING COUNT(*) >= 100
ORDER BY taxa_churn_pct DESC;
```

O Problema não é "Geográfico", é de "Engajamento Genérico": A taxa ligeiramente maior em Tỉnh khác (Outras Províncias) ocorre provavelmente porque zonas periféricas ou fora dos grandes centros possuem menor capilaridade de atendimento físico ou campanhas de engajamento direcionadas. No entanto, o motivo raiz do cancelamento continua sendo o mesmo de toda a base: baixo número de serviços utilizados (média de 2.4 / 2.5) e perda de atividade.
